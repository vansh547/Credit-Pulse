```python
import pandas as pd
import numpy as np
```

```python
data=r"C:\Users\Mahakaal\CreditPulse\data\raw\credit_score.csv"
df=pd.read_csv(data)
```

```python
df.head().T
```

```python
df.info()
```

```python
df=df.drop(columns=['Unnamed: 0'])
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.head().T
```

```python
df.isnull().mean()*100
```

```python
monthlyincome_median = df['monthlyincome'].median()
df['monthlyincome'] = df['monthlyincome'].fillna(monthlyincome_median)

numberofdependents_median = df['numberofdependents'].median()
df['numberofdependents'] = df['numberofdependents'].fillna(numberofdependents_median)
```
